```python
import numpy as np
import pandas as pd
```

##### Load Data

```python
df=pd.read_csv(r'anime.csv') # r lagane se path me '\' hota he use '\n' escape sequence jo ki new line ya '\t' tab nahi smjata
df.head()
```

```python
def extract_episode(txt):
    flag=False
    episodes=""
    for i in txt:
        if i=='(':
            flag=True
            continue
        if flag and i==' ':
            break
        if flag:
            episodes+=i
    return episodes
```

```python
def extract_date(txt):
    
    flag=False
    ans=""
    
    for ch in str(txt):
        
        if ch==')':
            flag=True
            continue
        
        if flag and ch==',':break
        if flag:ans+=ch
        
    last=ans.rfind(' ')
    
    if last!=-1:ans=ans[:last+5]
        
    return ans
```

```python
def get_total_months(time_str):
    time_str = str(time_str)
    
    if '-' not in time_str:
        return 0
        
    m_dict = {"Jan": 1, "Feb": 2, "Mar": 3, "Apr": 4, "May": 5, "Jun": 6, 
              "Jul": 7, "Aug": 8, "Sep": 9, "Oct": 10, "Nov": 11, "Dec": 12}
    
    try:
        parts = time_str.split('-')
        start = parts[0].strip()
        end = parts[1].strip()
        
        s_m, s_y = m_dict.get(start[:3], 0), int(start[-4:])
        e_m, e_y = m_dict.get(end[:3], 0), int(end[-4:])
        
        if s_m == 0 or e_m == 0:
            return 0
            
        total_months = (e_y - s_y) * 12 + (e_m - s_m)
        
        return total_months if total_months > 0 else 0
    except:
        return 0
```

```python
df['Episodes']=df['Title'].apply(extract_episode)
df['Episodes']=df['Episodes'].astype(int)
df['Total Time']=df['Title'].apply(extract_date)
df['Months']=df['Total Time'].apply(get_total_months)
df.head()
```

```python
# Higesh Score
highest_score=df[df['Score']==df['Score'].max()]
highest_score

# Ye sahi kyu chala?

# df['Score'] == df['Score'].max() ek condition check karta hai. Ye har ek row ke paas jata hai aur dekhta hai ki kya uska score highest hai?

# Iske result mein ek True aur False ki list banti hai (jis row ka score max hai wahan True, baaki sab jagah False). Isko "Boolean Mask" kehte hain.

# Jab tu df[True/False wali list] paas karta hai, toh Pandas samajh jata hai ki tujhe Rows filter karni hain. Isliye wo sirf unhi rows ko bahar nikal ke deta hai jahan condition True thi.

# Short mein:

# Pehle wale mein tu direct value (number) daal raha tha, jisko Pandas column samajh baitha.

# Doosre wale mein tu Condition (==) daal raha hai, jisse Pandas rows filter karke deta hai.
```
