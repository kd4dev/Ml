# Series

* A Pandas Series is a one-dimensional, labeled array capable of holding data of any type (integers, strings, floats, Python objects, etc.). You can think of it as a single column in an Excel spreadsheet or a database table.
* index are called as labels


```python
import pandas as pd
import numpy as np
```

```python
my_list=[1,2,3,34]
labels=['a','b','c','d']
my_arr=np.array([1,2,3,4,4])
my_dic={1:10,2:23,3:"kedar"}
```

```python
a=pd.Series(my_list) # Python list se series
b=pd.Series(my_arr) # numpay array se series
c=pd.Series(my_dic) # for dictionaries keys are their indexes
d=pd.Series(my_list,index=labels) #lenght of both should be same
print(a)
print(b)
print(c)
print(d)
```

## MISSING VALUES

```python
newData = {
    'A': [1, 2, np.nan, 4, 5],
    'B': [1, 2, 3, 4, 5],
    'C': [1, 2, 3, np.nan, np.nan],
    'D': [1, np.nan, np.nan, np.nan, 5]
}
df=pd.DataFrame(newData)
df.isna()
# A	B	C	D
# 0	False	True	False	False
# 1	False	False	False	True
# 2	True	False	False	True
# 3	False	False	True	True
# 4	False	False	True	False

df.isna().sum()
# 	0
# A	1
# B	1
# C	2
# D	3

df.isna().any() # ek bhi null hoga toh col ko true mark karega

#       0
# A	True
# B	True
# C	True
# D	True


```

```python
# row ke uppar chalta he
df2=df.dropna()
df2=df.dropna(thresh=3) # kam se kam 3 non null honi chahiye tab rahega
print(df) # df se kuch nahi hatega
print(df2)
```

```python
# dropna ----> row wise
# isna ----> col wise
# fillna ----> col wise

df.isna() # search
df.dropna() # remove

df.fillna(0) # filling null as zero
df.fillna(value=0) # pehli attribute default value hota he-->

			# print(df) # isme kuch fill na hoga naya df banana padega
			# inplace=true kar do naye df ki jarurat nahi padegi

val={'A':1,'B':2,'C':3,'D':4}
# val=[1,2,3,4] --->"value" parameter must be a scalar or dict,
df.fillna(df.mean())
df.fillna(value=val,inplace=True)
df
```

# Merging Joining Concatination

## MERGING

```python
employees = pd.DataFrame({
    'employee_id': [1, 2, 3, 4, 5],
    'name': ['John', 'Anna', 'Peter', 'Linda', 'Bob'],
    'department': ['HR', 'IT', 'Finance', 'IT', 'HR']
})

salaries = pd.DataFrame({
    'employee_id': [1, 2, 3, 6, 7],
    'salary': [60000, 80000, 65000, 70000, 90000],
    'bonus': [5000, 10000, 7000, 8000, 12000]
})  					 
```

```python
a1=pd.merge(employees,salaries,on='employee_id',how='inner') # inner join type-->common included
a2=pd.merge(employees,salaries,on='employee_id',how='outer') # outer join type -->all included
a3=pd.merge(employees,salaries,on='employee_id',how='left') # left join type-->left wale ke sare ayenge
a4=pd.merge(employees,salaries,on='employee_id',how='right') # right join type-->right wale ke sare ayenge
print(employees)
print(salaries)
# new df return karega
```

## CONCATINATION

```python
df1 = pd.DataFrame({
    'A': ['A0', 'A1', 'A2'],
    'B': ['B0', 'B1', 'B2'],
    'C': ['C0', 'C1', 'C2']
})
df2 = pd.DataFrame({
    'A': ['A3', 'A4', 'A5'],
    'B': ['B3', 'B4', 'B5'],
    'C': ['C3', 'C4', 'C5']
})
```

```python
pd.concat([df1,df2])
print(df1)
print(df2)
# new df return karega
pd.concat([df1,df2],axis=1) # row wise
```

## JOINING

```python
df1 = pd.DataFrame({
    'name': ['Alice', 'Bob', 'Charlie']
}, index=[1, 2, 3])

# Second DataFrame
df2 = pd.DataFrame({
    'score': [85, 90, 75]
}, index=[2, 3, 4])
```

# Group By 
* Group by object return krta he

```python
data1 = {
    'Category': ['A', 'B', 'A', 'B', 'A', 'B', 'A', 'B'],
    'Store': ['S1', 'S1', 'S2', 'S2', 'S1', 'S2', 'S2', 'S1'],
    'Sales': [100, 200, 150, 250, 120, 180, 200, 300],
    'Quantity': [10, 15, 12, 18, 8, 20, 15, 25],
    'Date': pd.date_range('2023-01-01', periods=8)
}
df1=pd.DataFrame(data1)
df1
```

```python
cat=df1.groupby('Category')
print(type(cat)) # <class 'pandas.core.groupby.generic.DataFrameGroupBy'>
print(cat) # <pandas.core.groupby.generic.DataFrameGroupBy object at 0x7a7759399f10>
# object ko direct print nahi kar skte
for category,v in cat:
    print('Category:'+category)
    print(v)
    
print(cat['Sales'].sum())
# Category
# A    570
# B    930
# Name: Sales, dtype: int64
```

### Group by multiple categories

```python
dog=df1.groupby(['Category','Store'])
for i,v in dog:
    print(i)
    print(v)
    print()
    print()

print(dog['Sales'].sum())

# Category  Store
#       A            S1           220
#                     S2           350
#       B            S1      500
#                      S2          430
# Name: Sales, dtype: int64
```

## AGGREGATION
* columns ke sare values ko combine krke kuch methods lagana jaise mean medain mode

```python
df1['Sales'].mean()
df1['Sales'].agg(['mean','max','min','median','std']) # mode nahi kar skte bas
```

# Pivot tables
* heat maps banande me use ayega

```python
data2 = {
    'Date': pd.date_range('2023-01-01', periods=20),
    'Product': ['A', 'B', 'C', 'D'] * 5,
    'Region': ['East', 'West', 'North', 'South', 'East', 'West', 'North', 'South', 'East', 'West',
               'North', 'South', 'East', 'West', 'North', 'South', 'East', 'West', 'North', 'South'],
    'Sales': np.random.randint(100, 1000, 20),
    'Units': np.random.randint(10, 100, 20),
    'Rep': ['John', 'Mary', 'Bob', 'Alice', 'John', 'Mary', 'Bob', 'Alice', 'John', 'Mary',
            'Bob', 'Alice', 'John', 'Mary', 'Bob', 'Alice', 'John', 'Mary', 'Bob', 'Alice']
}

df2 = pd.DataFrame(data2)

# print(df2)
df2['Month'] = df2['Date'].dt.month_name()
# print(df2)
df2['Quarter'] = 'Q' + df2['Date'].dt.quarter.astype(str)
# print(df2)
```

```python
xy=df2.pivot_table(index='Region',values='Sales',columns='Product') # values me mean ayega
# print(type(xy)) # <class 'pandas.core.frame.DataFrame'> 
xy
```

```python
# pivot2=df2.pivot_table(index='Region',columns='Product',values=['Sales','Units'])
# pivot2
df2.pivot_table(index='Region', columns='Product', values='Sales', aggfunc='mean')
```

```python
pd.crosstab(df2['Region'], df2['Product']) # isme mean nahi hoga,count hoga
```
