# Pandas: Series and DataFrame

Pandas Python ki popular data-analysis library hai. Is notebook mein Series aur DataFrame se lekar filtering, missing values, combining data, grouping aur pivot tables tak ke important concepts examples ke saath cover kiye gaye hain.

## How to read the comments

Har code cell ke upar comments mein `Returns`, `Works on`, `Does not work` aur `Common error` diya gaya hai. Pandas methods aksar original DataFrame ko modify nahi karte; woh naya object return karte hain, jab tak `inplace=True` explicitly na diya ho.

## Series kya hoti hai?

- `Series` ek one-dimensional labelled array hoti hai.
- Har value ke saath ek **index/label** hota hai.
- Isse ek Excel column ya SQL table ke ek column ki tarah samajh sakte hain.
- Series list, NumPy array ya dictionary se ban sakti hai.

```python
# Returns: None (imports/configuration; display settings change Pandas output only)
# Works on: Python environment with pandas and numpy installed.
# Common error: ModuleNotFoundError if libraries are not installed.

# Required libraries
import numpy as np
import pandas as pd

pd.set_option('display.max_columns', None)
pd.set_option('display.width', 100)
```

```python
# Returns: list, numpy.ndarray and dict objects; no Pandas object yet.
# Works on: normal Python sequences.
# Common error: array/list length mismatch later when assigning custom labels.

# Series banane ke liye teen common input structures
my_list = [1, 2, 3, 34]
labels = ['a', 'b', 'c', 'd']
my_array = np.array([1, 2, 3, 4, 4])
my_dict = {1: 10, 2: 23, 3: 'kedar'}
```

```python
# Returns: pandas.Series in every case; dictionary keys become the Series index.
# Works on: list, NumPy array and dictionary.
# Does not work: custom index length must match the number of values.
# Common error: ValueError when index and values have different lengths.

# List, NumPy array, dictionary aur custom index se Series
series_from_list = pd.Series(my_list)
series_from_array = pd.Series(my_array)
series_from_dict = pd.Series(my_dict)
series_with_labels = pd.Series(my_list, index=labels)

print('From list:\n', series_from_list, sep='')
print('\nFrom NumPy array:\n', series_from_array, sep='')
print('\nFrom dictionary (keys become index):\n', series_from_dict, sep='')
print('\nWith custom labels:\n', series_with_labels, sep='')
```

# DataFrame

DataFrame ek two-dimensional, tabular data structure hai—bilkul Excel sheet ya SQL table jaisa. Ismein rows aur columns dono ke labels hote hain.

```python
# Returns: pandas.DataFrame; dictionary keys become column names.
# Works on: dictionary values must have equal lengths (or be scalars).
# Common error: ValueError: All arrays must be of the same length.

# Dictionary ki keys columns aur lists rows banati hain
data = {
    'Name': ['John', 'Anna', 'Peter', 'Linda'],
    'Age': [28, 34, 29, 42],
    'City': ['New York', 'Paris', 'Berlin', 'London'],
    'Salary': [65000, 70000, 62000, 85000]
}

df_from_dict = pd.DataFrame(data)
df_from_dict
```

```python
# Returns: pandas.DataFrame. Without columns=, labels are 0, 1, 2...
# Works on: list of rows; each row should have the same number of values.
# Common error: column-label count must equal values per row.

# List of lists se DataFrame
data_list = [
    ['John', 28, 'New York', 65000],
    ['Anna', 34, 'Paris', 70000],
    ['Peter', 29, 'Berlin', 62000],
    ['Linda', 42, 'London', 85000]
]

df_default_columns = pd.DataFrame(data_list)
df_named_columns = pd.DataFrame(
    data_list, columns=['Name', 'Age', 'City', 'Salary']
)

print('Default column labels:')
display(df_default_columns)
print('Named columns:')
display(df_named_columns)
```

```python
# Single [] selection returns a Series; double [[]] selection returns a DataFrame.
# Works on: existing column labels.
# Common error: KeyError when a requested column name does not exist.

# Column selection: single column Series deta hai, multiple columns DataFrame
print('Single column:\n', df_named_columns['Name'], sep='')
print('\nMultiple columns:')
display(df_named_columns[['Name', 'Age']])
```

```python
# Assignment returns None; the DataFrame is modified by adding a column.
# Works on: scalar values or list-like values whose length equals row count.
# Common error: ValueError when a list has the wrong length.

# Naya column add karna
df_named_columns['Experience'] = [3, 8, 4, 15]
df_named_columns
```

```python
# drop() returns a new DataFrame; original data remains unchanged by default.
# Works on: row labels with index= or columns with columns=.
# Common errors: KeyError for a missing label; use inplace=True only when mutation is intended.

# drop() default roop se naya DataFrame return karta hai; original tabhi badlega jab inplace=True ho
without_experience = df_named_columns.drop(columns='Experience')
print('Without Experience:')
display(without_experience)
print('Original DataFrame unchanged:')
display(df_named_columns)

# Rows drop karne ke liye axis=0 ya index= use kar sakte hain
rows_removed = df_named_columns.drop(index=[0, 1])
rows_removed
```

```python
# loc returns Series/scalar/DataFrame depending on selection; iloc follows integer positions.
# loc: label-based and label slice stop is included.
# iloc: position-based and slice stop is excluded.
# Common errors: KeyError with wrong labels; IndexError with invalid positions.

# loc = labels ke basis par; iloc = integer positions ke basis par
print('Row label 0:\n', df_named_columns.loc[0], sep='')
print('\nRows 0 and 1:')
display(df_named_columns.loc[[0, 1]])
print('Row at position 3:\n', df_named_columns.iloc[3], sep='')
print('\nValue at row label 1, column Age:', df_named_columns.loc[1, 'Age'])

print('\niloc slicing (stop excluded):')
display(df_named_columns.iloc[1:3, 1:3])
print('loc slicing (label stop included):')
display(df_named_columns.loc[1:2, 'Age':'City'])
```

```python
# Boolean filtering returns a DataFrame. Conditions must produce a Boolean Series.
# Works on: comparable columns; use &, | and ~ for multiple conditions.
# Common error: use parentheses around every condition; Python 'and'/'or' do not work element-wise.

# Filtering mein har condition ko parentheses mein likhna zaroori hai
print('Selected rows and columns:')
display(df_named_columns.loc[[0, 1], ['City', 'Salary']])
print('Age greater than 30:')
display(df_named_columns[df_named_columns['Age'] > 30])
print('Age > 30 and city is Paris:')
display(df_named_columns[(df_named_columns['Age'] > 30) & (df_named_columns['City'] == 'Paris')])
```

# Missing Values

Real-world data mein missing values common hoti hain. Pandas mein inhe generally `NaN` se represent kiya jata hai. Pehle missing values identify karein, phir context ke according remove ya fill karein.

```python
# isna() returns a Boolean DataFrame; sum() returns missing counts as a Series; any() returns Boolean Series.
# Works on: DataFrames/Series with NaN, None or NaT values.
# Common error: missing values may have different representations, so check with isna().

new_data = {
    'A': [1, 2, np.nan, 4, 5],
    'B': [1, 2, 3, 4, 5],
    'C': [1, 2, 3, np.nan, np.nan],
    'D': [1, np.nan, np.nan, np.nan, 5]
}
missing_df = pd.DataFrame(new_data)

print('Missing-value mask:')
display(missing_df.isna())
print('\nMissing values per column:')
print(missing_df.isna().sum())
print('\nDoes each column contain at least one missing value?')
print(missing_df.isna().any())
```

```python
# dropna() returns a new DataFrame with rows removed. thresh=3 keeps rows with >=3 non-null values.
# Works on: rows by default; use axis=1 for columns.
# Common error: dropping all rows when missing data is extensive—inspect isna().sum() first.

# dropna() rows remove karta hai; thresh minimum non-null values define karta hai
complete_rows = missing_df.dropna()
minimum_three_values = missing_df.dropna(thresh=3)

print('Rows with no missing values:')
display(complete_rows)
print('Rows with at least 3 non-null values:')
display(minimum_three_values)
```

```python
# fillna() returns a new DataFrame with missing values replaced.
# It accepts a scalar, Series, or column-to-value dictionary.
# Common error: mean filling works only for numeric columns; use numeric_only=True where needed.

# Missing values fill karne ke options
filled_with_zero = missing_df.fillna(0)
filled_with_mean = missing_df.fillna(missing_df.mean(numeric_only=True))

# Column-specific values ke liye dictionary useful hai
fill_values = {'A': 1, 'B': 2, 'C': 3, 'D': 4}
filled_by_column = missing_df.fillna(value=fill_values)

print('Filled with zero:')
display(filled_with_zero)
print('Filled with column means:')
display(filled_with_mean)
print('Filled with custom values:')
display(filled_by_column)

# Note: fillna() original DataFrame ko change nahi karta jab tak inplace=True na diya ho.
```

# Combining DataFrames

Pandas mein alag-alag tables ko combine karne ke teen common tareeqe hain:

1. **Merge** — common key/column ke basis par SQL-style join.
2. **Concat** — rows ya columns ko stack karna.
3. **Join** — index ke basis par combine karna.

## 1. Merge

`pd.merge()` common key ke basis par do DataFrames ko combine karta hai. `how` se join type choose hota hai: `inner`, `outer`, `left` ya `right`.

```python
# Returns: two pandas.DataFrames used as merge inputs; no join happens in this cell.
# Data requirement: merge key should exist in both tables for on='employee_id'.

employees = pd.DataFrame({
    'employee_id': [1, 2, 3, 4, 5],
    'name': ['John', 'Anna', 'Peter', 'Linda', 'Bob'],
    'department': ['HR', 'IT', 'Finance', 'IT', 'HR']
})

salaries = pd.DataFrame({
    'employee_id': [1, 2, 3, 6, 7],
    'salary': [60000, 80000, 65000, 70000, 90000],
    'bonus': [5000, 10000, 7000, 8000, 12000]
})  					 
```

```python
# pd.merge() returns a DataFrame. It performs SQL-style key-based joins.
# inner=matching keys only; outer=all keys; left=all left keys; right=all right keys.
# Common errors: KeyError for missing key; duplicate keys can multiply rows.

# Join types ka result compare karein
inner_merge = pd.merge(employees, salaries, on='employee_id', how='inner')
outer_merge = pd.merge(employees, salaries, on='employee_id', how='outer')
left_merge = pd.merge(employees, salaries, on='employee_id', how='left')
right_merge = pd.merge(employees, salaries, on='employee_id', how='right')

print('Inner merge (sirf matching IDs):')
display(inner_merge)
print('Outer merge (dono tables ke saare IDs):')
display(outer_merge)
```

## 2. Concatenation

`pd.concat()` DataFrames ko axis ke along jodta hai:

- `axis=0`: rows append/stack hoti hain.
- `axis=1`: columns side-by-side aate hain.

```python
# Returns: two DataFrames with compatible columns for row-wise concatenation.
# For axis=0, columns are aligned by name; missing columns produce NaN.

df1 = pd.DataFrame({
    'A': ['A0', 'A1', 'A2'],
    'B': ['B0', 'B1', 'B2'],
    'C': ['C0', 'C1', 'C2']
})
df2 = pd.DataFrame({
    'A': ['A3', 'A4', 'A5'],
    'B': ['B3', 'B4', 'B5'],
    'C': ['C3', 'C4', 'C5']
})
```

```python
# pd.concat() returns a DataFrame. axis=0 stacks rows; axis=1 places columns side-by-side.
# ignore_index=True creates a fresh continuous row index.
# Common error: axis=1 aligns by index, so mismatched indexes may create NaN.

print('Row-wise concatenation:')
display(pd.concat([df1, df2], ignore_index=True))
print('Column-wise concatenation:')
display(pd.concat([df1, df2], axis=1))
```

## 3. Join

`DataFrame.join()` default roop se index ke basis par left join karta hai. `how='outer'` use karne par dono DataFrames ke saare index preserve hote hain.

```python
# Returns: two DataFrames whose indexes are intentionally different.
# join() will match rows using these indexes, not a normal column key.

df1 = pd.DataFrame({
    'name': ['Alice', 'Bob', 'Charlie']
}, index=[1, 2, 3])

# Second DataFrame
df2 = pd.DataFrame({
    'score': [85, 90, 75]
}, index=[2, 3, 4])
```

```python
# join() returns a DataFrame and is index-based. Default is left join; how='outer' keeps all indexes.
# Common error: overlapping column names need lsuffix/rsuffix.

print('Default left join:')
display(df1.join(df2))
print('Outer join:')
display(df1.join(df2, how='outer'))
```

# GroupBy

`groupby()` data ko groups mein divide karta hai. Iske baad har group par aggregation jaise `sum()`, `mean()` ya `count()` laga sakte hain. `groupby()` khud result table nahi, balki ek GroupBy object return karta hai.

```python
# Returns: DataFrame containing raw sales data and a datetime column.
# pd.date_range() returns a DatetimeIndex; periods controls its length.
# Common error: date strings may need pd.to_datetime() before using .dt.

data1 = {
    'Category': ['A', 'B', 'A', 'B', 'A', 'B', 'A', 'B'],
    'Store': ['S1', 'S1', 'S2', 'S2', 'S1', 'S2', 'S2', 'S1'],
    'Sales': [100, 200, 150, 250, 120, 180, 200, 300],
    'Quantity': [10, 15, 12, 18, 8, 20, 15, 25],
    'Date': pd.date_range('2023-01-01', periods=8)
}
df1=pd.DataFrame(data1)
df1
```

```python
# groupby() returns DataFrameGroupBy, not a final DataFrame.
# It works on one or more categorical/grouping columns; aggregation is needed for a summary.
# Common error: grouping by a misspelled column raises KeyError.

category_groups = df1.groupby('Category')
print('Object type:', type(category_groups))

for category, group in category_groups:
    print(f'Category: {category}')
    display(group)

print('Total sales by category:')
print(category_groups['Sales'].sum())
```

## Multiple columns par GroupBy

Ek se zyada columns dene par hierarchical grouping hoti hai—for example, pehle `Category`, phir `Store`.

```python
# Multiple-column groupby returns DataFrameGroupBy; aggregation returns a Series with MultiIndex.
# Works on combinations of Category and Store.
# Common error: remember that tuple keys such as ('A', 'S1') identify groups.

category_store_groups = df1.groupby(['Category', 'Store'])

for group_name, group in category_store_groups:
    print(f'Group: {group_name}')
    display(group)

print('Sales by category and store:')
print(category_store_groups['Sales'].sum())
```

## Aggregation

Aggregation ka matlab hai multiple values ko summarize karna—jaise mean, median, minimum, maximum ya standard deviation.

```python
# mean() returns a scalar; agg([...]) returns a Series of summary statistics.
# Works on numeric Series.
# Common error: mean/median on non-numeric text columns may fail or be meaningless.

print('Mean sales:', df1['Sales'].mean())
print('Multiple statistics:')
df1['Sales'].agg(['mean', 'max', 'min', 'median', 'std'])
```

# Pivot Tables

Pivot table grouped data ka compact summary banati hai. Ye reporting aur heatmaps ke liye useful hoti hai. `aggfunc` default roop se `mean` hota hai; zarurat ke hisaab se `sum`, `count`, `max` etc. use kar sakte hain.

```python
# Returns: DataFrame; .dt.month_name() and .dt.quarter return Series.
# Works on datetime-like Date values. rng makes random output reproducible.
# Common error: .dt accessor fails when the column is still plain text.

# Demo data ko reproducible banane ke liye random seed set kiya gaya hai
rng = np.random.default_rng(42)
data2 = {
    'Date': pd.date_range('2023-01-01', periods=20),
    'Product': ['A', 'B', 'C', 'D'] * 5,
    'Region': ['East', 'West', 'North', 'South'] * 5,
    'Sales': rng.integers(100, 1000, 20),
    'Units': rng.integers(10, 100, 20),
    'Rep': ['John', 'Mary', 'Bob', 'Alice'] * 5
}

pivot_source = pd.DataFrame(data2)
pivot_source['Month'] = pivot_source['Date'].dt.month_name()
pivot_source['Quarter'] = 'Q' + pivot_source['Date'].dt.quarter.astype(str)
pivot_source.head()
```

```python
# pivot_table() returns a DataFrame; default aggregation is mean when duplicates exist.
# index defines rows, columns defines columns, values defines measured data.
# Common error: pivot_table and pivot differ—pivot fails when index/column pairs are duplicated.

# Region rows, Product columns; har cell mein average Sales
sales_pivot = pivot_source.pivot_table(
    index='Region', columns='Product', values='Sales', aggfunc='mean'
)
sales_pivot
```

```python
# pivot_table() returns a DataFrame, here with MultiIndex columns for Sales and Units.
# aggfunc can be a string or a dictionary mapping each value column to its method.

# Multiple values aur custom aggregation bhi possible hai
pivot_summary = pivot_source.pivot_table(
    index='Region',
    columns='Product',
    values=['Sales', 'Units'],
    aggfunc={'Sales': 'sum', 'Units': 'mean'}
)
pivot_summary
```

```python
# crosstab() returns a DataFrame of frequency counts.
# It counts combinations; it does not calculate mean or sum unless values/aggfunc are supplied.
# Common error: passing non-aligned Series can produce unexpected counts.

# crosstab frequency/count table banata hai, average nahi
pd.crosstab(pivot_source['Region'], pivot_source['Product'])
```

# DataFrame: Basic Operations

Is section mein shape, columns, data types, summary statistics, column transformation aur `apply()` ka use dekhenge.

```python
# Returns: a numeric DataFrame for inspection and transformation practice.

dfxy = pd.DataFrame({
    'A': [1, 2, 3, 4, 5],
    'B': [10, 20, 30, 40, 50],
    'C': [100, 200, 300, 400, 500]
})
```

```python
# shape returns a tuple; columns returns an Index; info() returns None but prints metadata.
# describe() returns a statistical summary DataFrame for numeric columns.
# apply() returns a transformed Series; assignment stores it as a new/replaced column.
# Common error: apply functions should handle the input dtype they receive.

# Dataset ka structure aur summary
print('Shape:', dfxy.shape)
print('Columns:', dfxy.columns.tolist())
print('\nInfo:')
dfxy.info()
print('\nDescriptive statistics:')
display(dfxy.describe())

# Vectorized operation: poore column par ek saath operation
dfxy['A'] = dfxy['A'] + 10

# apply() custom function ya lambda apply karne ke liye
def square(value):
    return value ** 2

dfxy['B'] = dfxy['B'].apply(square)
dfxy['D'] = dfxy['B'].apply(lambda value: value * 3)
dfxy
```

# Quick Revision & Best Practices

- `loc` labels ke liye aur `iloc` positions ke liye use karein.
- Filtering conditions ko parentheses mein rakhein; multiple conditions ke liye `&` / `|` use karein.
- `drop()` aur `fillna()` by default naya DataFrame return karte hain—original data safe rehta hai.
- `merge()` key-based, `concat()` axis-based aur `join()` index-based combining ke liye yaad rakhein.
- Random demo data ke liye seed set karna reproducible results deta hai.
- Large datasets mein unnecessary `inplace=True` se bachna aur method chaining ko readable rakhna useful hota hai.

```python
# head() returns a DataFrame; dtypes returns a Series; unique() returns a NumPy array.
# value_counts() returns a frequency Series; sort_values() returns a sorted DataFrame.
# duplicated().sum() returns an integer count.
# Common error: sorting by a missing column raises KeyError.

# Common inspection checklist
print('First rows:')
display(pivot_source.head())
print('\nData types:')
print(pivot_source.dtypes)
print('\nUnique products:', pivot_source['Product'].unique())
print('\nRows per region:')
print(pivot_source['Region'].value_counts())

# Sorting aur duplicate check bhi data cleaning mein common hain
sorted_sales = pivot_source.sort_values('Sales', ascending=False)
print('\nTop 5 sales records:')
display(sorted_sales.head())
print('Duplicate rows:', pivot_source.duplicated().sum())
```

# Method Return-Type Cheat Sheet

| Method / operation | Typical return | Main use |
|---|---|---|
| `df['col']` | `Series` | Single column select |
| `df[['a', 'b']]` | `DataFrame` | Multiple columns select |
| `loc` / `iloc` | scalar, `Series`, or `DataFrame` | Label/position selection |
| `isna()` | Boolean `DataFrame`/`Series` | Missing-value detection |
| `dropna()` / `fillna()` | New `DataFrame`/`Series` | Remove/fill missing values |
| `merge()` / `concat()` / `join()` | `DataFrame` | Combine data |
| `groupby()` | `DataFrameGroupBy` | Create groups before aggregation |
| `sum()`, `mean()` | scalar or `Series` | Summarize values |
| `agg()` | `Series` or `DataFrame` | Multiple summaries |
| `pivot_table()` / `crosstab()` | `DataFrame` | Summary/report tables |
| `apply()` | Usually `Series` | Apply custom function |

**Rule of thumb:** method ka result dekhne ke liye usse variable mein save karein ya cell ke last line par likhein. `print()` ka return value `None` hota hai; woh sirf output screen par display karta hai.
