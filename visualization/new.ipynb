```python
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
from random import sample
```

```python
x=np.linspace(1,10,10)
y=x**2
print(x)
print()
print(y)
```

```python
plt.plot(x,y) # x=x,y=y bhi de skte ho par default pehela x aur dusra y axis hi hota he
plt.title("Quadratic")
plt.xlabel("x-axis")
plt.ylabel("y-axis")
```

```python
# figure se canvas set hua
# fig.add se ek graph set hua
#.plot se us graph per plot kiya

fig=plt.figure(figsize=(4,4),dpi=1600) # figsize=600*800(width*height)
plot1=fig.add_axes([0.1,0.1,1,1])
plot2=fig.add_axes([0.2,0.55,0.5,0.5])

plot1.plot(x,y)
plot2.plot(x,y)
```
